```python
import numpy as np
import matplotlib.pyplot as plt

Emin = []
Emax = []
proton = []
neutron = []

with open("spectrum_lead.out") as f:
    for line in f:

        cols = line.split()

        # asegurarse de que la línea tenga datos numéricos
        if len(cols) >= 6:
            try:
                Emin.append(float(cols[0]))
                Emax.append(float(cols[1]))
                proton.append(float(cols[2]))
                neutron.append(float(cols[4]))
            except ValueError:
                continue

Emin = np.array(Emin)
Emax = np.array(Emax)
proton = np.array(proton)
neutron = np.array(neutron)

Ecenter = (Emin + Emax)/2

plt.step(Ecenter, proton, where="mid", label="proton")
plt.step(Ecenter, neutron, where="mid", label="neutron")

plt.yscale("log")

plt.xlabel("Energy (MeV)")
plt.ylabel("Flux [1/cm²/source]")
plt.title("Spectrum in lead plate")

plt.legend()
plt.show()
```
